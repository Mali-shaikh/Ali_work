```python
import zipfile

# Ensure rarfile is installed before importing
!pip install rarfile
import rarfile

# Specify the path to your .rar file and the extraction directory
rar_file_path = "/content/mask_data.zip"
extraction_path = "/content/extracted_dataset"

# Create the extraction directory if it doesn't exist
import os
if not os.path.exists(extraction_path):
    os.makedirs(extraction_path)

# Open the .rar file and extract its contents
try:
    with rarfile.RarFile(rar_file_path, 'r') as rf:
        rf.extractall(extraction_path)
    print(f"Successfully extracted '{rar_file_path}' to '{extraction_path}'")
except rarfile.BadRarFile:
    print(f"Error: The file '{rar_file_path}' is not a valid RAR archive. Attempting with zipfile...")
    try:
        with zipfile.ZipFile(rar_file_path, 'r') as zf:
            zf.extractall(extraction_path)
        print(f"Successfully extracted '{rar_file_path}' as a ZIP file to '{extraction_path}'")
    except zipfile.BadZipFile:
        print(f"Error: The file '{rar_file_path}' is neither a valid RAR nor a valid ZIP archive.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
